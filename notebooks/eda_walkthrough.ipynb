# AI Candidate Scoring: Exploratory Data Analysis & Historical Bias Audit
This notebook demonstrates exploratory data analysis, merit correlations, and audits synthetic historical bias across demographic groups (Gender, Age, College Tier).

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv('../data/processed/cleaned_resumes.csv')
print(f"Loaded {len(df):,d} candidates across {df['job_role_applied'].nunique()} technical roles.")
df.head()
```

## 1. Class Distribution
Analyzing class balance between shortlisted and rejected candidates.

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='shortlisted', palette=['#EF4444', '#10B981'])
plt.title('Shortlisted (1) vs Rejected (0) Distribution')
plt.show()
```

## 2. Demographic & Historical Disparities
Checking raw selection rates by gender, age group, and college tier.

```python
df['age_group'] = pd.cut(df['age'], bins=[0, 29, 45, 100], labels=['<30', '30-45', '45+'])
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), sharey=True)
sns.barplot(data=df, x='gender', y='shortlisted', ax=axes[0], palette='Blues')
axes[0].set_title('Shortlist Rate by Gender')
sns.barplot(data=df, x='age_group', y='shortlisted', ax=axes[1], palette='Greens')
axes[1].set_title('Shortlist Rate by Age Group')
sns.barplot(data=df, x='college_tier', y='shortlisted', ax=axes[2], palette='Oranges')
axes[2].set_title('Shortlist Rate by College Tier')
plt.suptitle('Raw Disparities in Historical Target Data')
plt.show()
```
